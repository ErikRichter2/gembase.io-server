```python
from src.server.models.dms_constants import DmsConstants

platform = DmsModel.get_dms_data_to_json(conn, guid=DmsConstants.platform_guid)
prompts = DmsModel.get_dms_data_to_json(conn, guid=DmsConstants.prompts_guid)
```

```python
def format_name(name: str) -> str:
    return name.replace("-", "_").replace(" ", "_").lower()

def prompt_tag_to_interface(prompt_tag: str):
    for it in platform["Interface"]:
        node = it["Product_node"]
        if node == "Solutions" and prompt_tag == "Solutions":
            pass
        if node == prompt_tag and it["Search_node"] is not None and it["Search_node"] != "":
            return it
    return None

final_data = {
    "existing_cols": {
        "topics": [],
        "genres": []
    },
    "potential_cols": {
        "topics": [],
        "genres": []
    },
    "items": {

    }
}
areas = pp.get_private_data()["google"]["google_docs"]["prompts"]["areas"]
for area in areas:
    for it in prompts[area]:
        is_genre = area == "Genres"
        tag = it["Tag"]
        interface = prompt_tag_to_interface(it["Tag"])
        if interface is None:
            continue
        if interface["Category"] == "Genre" and not is_genre:
            continue
        if interface["Category"] == "Topic" and is_genre:
            continue

        c = interface["Category"]
        s = interface["Subcategory"]
        n = interface["Search_node"]
        def_id = f"{GembaseUtils.name_to_col(c)}__{GembaseUtils.name_to_col(s)}__{GembaseUtils.name_to_col(n)}"

        search_node = interface["Search_node"]
        if interface is not None:
            tagged_col = f"{format_name(area)}__{format_name(it['Category'])}__{format_name(tag)}"

            try:
                row = conn.select_one_or_none(f"""
                SELECT ROUND((SUM(a.downloads_us_gp_12m) / 6) / 155000000 * 8, 2) AS product
                  FROM tagged_data.apps tg,
                       scraped_data.gp_ext_data a
                 WHERE tg.{tagged_col} = 1
                   AND tg.app_id = a.app_id
                """)
                if row is not None:
                    if search_node not in final_data["items"]:
                        final_data["items"][search_node] = {
                            "id": def_id,
                            "prompts": 0,
                            "survey": 0,
                            "def": interface,
                            "potential_id": "",
                            "existing_id": "",
                        }
                    final_data["items"][search_node]["prompts"] = float(row['product'])
                    final_data["items"][search_node]["existing_id"] = tagged_col
                    if is_genre:
                        final_data["existing_cols"]["genres"].append({
                            "col": tagged_col,
                            "def_id": def_id
                        })
                    else:
                        final_data["existing_cols"]["topics"].append({
                            "col": tagged_col,
                            "def_id": def_id
                        })
            except Exception as err:
                print(f"NOT FOUND: Tag: {tag} -> Search Node: {search_node} -> {tagged_col}")
```

```python
def find_def(gr: str, c: str):
    res = []
    p = sc.get_prefix(gr)
    c = c.replace(p, "")
    for it in SURVEY_DEF[gr]:
        if it["id"] == c:
            for interface in platform["Interface"]:
                if interface["Audience_node"].lower() == it["name"].lower():
                    if interface["Search_node"] is not None and interface["Search_node"] != "":
                        res.append(interface)
    return res

cnt = len(df.index)
for col in df.columns:
    for k in SURVEY_DEF:
        p = sc.get_prefix(k)
        if p in col:
            for d in find_def(k, col):
                c = d["Category"]
                s = d["Subcategory"]
                n = d["Search_node"]
                def_id = f"{GembaseUtils.name_to_col(c)}__{GembaseUtils.name_to_col(s)}__{GembaseUtils.name_to_col(n)}"

                search_node = d["Search_node"]
                if search_node not in final_data["items"]:
                    final_data["items"][search_node] = {
                        "id": def_id,
                        "prompts": 0,
                        "survey": 0,
                        "def": d,
                        "potential_id": "",
                        "existing_id": ""
                    }
                final_data["items"][search_node]["survey"] = round(len(df[df[col] >= 0.75].index) / cnt * 100, 2)
                final_data["items"][search_node]["potential_id"] = col
                if k == sc.GENRES or k == sc.GENRES_SUB:
                    final_data["potential_cols"]["genres"].append({
                        "def_id": def_id,
                        "col": col,
                    })
                elif k == sc.TOPICS or k == sc.THEMES:
                    final_data["potential_cols"]["topics"].append({
                        "def_id": def_id,
                        "col": col,
                    })
```

```python

dms_data = []
for k in final_data["items"]:
    potential = final_data["items"][k]["survey"]
    existing = final_data["items"][k]["prompts"]
    value = round(((potential - existing) / 100) * 155000000)
    if value < 1000:
        value = round(value / 100) * 100
    elif value < 10000:
        value = round(value / 1000) * 1000
    elif value < 100000:
        value = round(value / 10000) * 10000
    else:
        value = round(value / 100000) * 100000

    c = final_data["items"][k]["def"]["Category"]
    s = final_data["items"][k]["def"]["Subcategory"]
    n = final_data["items"][k]["def"]["Search_node"]
    potential_id = final_data["items"][k]["potential_id"]
    existing_id = final_data["items"][k]["existing_id"]
    if existing_id != "" and potential_id != "":
        dms_data.append({
            "id": f"{GembaseUtils.name_to_col(c)}__{GembaseUtils.name_to_col(s)}__{GembaseUtils.name_to_col(n)}",
            "category": c,
            "subcategory": s,
            "node": n,
            "value": value,
            "existing_id": existing_id,
            "potential_id": potential_id
        })

DmsModel.save_json_to_dms(conn, {
    "items": dms_data,
    "existing_cols": final_data["existing_cols"],
    "potential_cols": final_data["potential_cols"]
}, guid=DmsConstants.search_gaps, file_type="search gaps")
conn.commit()
```
