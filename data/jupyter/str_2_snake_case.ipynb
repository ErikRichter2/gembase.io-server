```python
from re import sub

def snake_case(s):
    # Replace hyphens with spaces, then apply regular expression substitutions for title case conversion
    # and add an underscore between words, finally convert the result to lowercase
    return '_'.join(
        sub('([A-Z][a-z]+)', r' \1',
        sub('([A-Z]+)', r' \1',
        s.replace('-', ' '))).split()).lower()

f = open("tmp.txt", "r")
ff = open("tmp2.txt", "w")
to_remove = ["!", ":", "&", "/", "?", "(", ")", "'", "*", ".", "-", ","]
for x in f:
    for r in to_remove:
        x = x.replace(r, "")
    x = x.replace("IP Type", "ip_type_").replace("IP", "ip_")
    ff.writelines(f"{snake_case(x)}\n")

ff.close()
```
