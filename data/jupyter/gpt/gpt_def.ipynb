```python
from src.db.mysql import DbConnection

prompts = [
    {"id": 1, "prompt_id": "gpt4_genres", "order_by": 1, "active": True},
    {"id": 2, "prompt_id": "gpt4_mechanics", "order_by": 1, "active": True},
    {"id": 3, "prompt_id": "gpt4_themes", "order_by": 1, "active": True},
    {"id": 4, "prompt_id": "gpt4_content", "order_by": 1, "active": True},
    {"id": 5, "prompt_id": "gpt4_genres_topics_prio", "order_by": 2, "active": True, "rank_prompt": True},
    {"id": 6, "prompt_id": "gpt4_themes_no_topics", "order_by": 1},
    {"id": 7, "prompt_id": "gpt4_topics_only", "order_by": 1}
]

conn = DbConnection()
conn.query("""
DELETE FROM tagged_data.def_prompts
""")
conn.query("""
DELETE FROM tagged_data.active_prompts_b
""")
sum_b = 0
for p in prompts:
    b = 1 << (p["id"] - 1)
    active = False if "active" not in p else p["active"]
    rank_prompt = False if "rank_prompt" not in p else p["rank_prompt"]
    if active:
        sum_b += b
    conn.query("""
    INSERT INTO tagged_data.def_prompts (id, prompt_id, order_by, active, b, rank_prompt)
    VALUES (%s, %s, %s, %s, %s, %s)
    """, [p["id"], p["prompt_id"], p["order_by"], active, b, rank_prompt])
conn.query("""
INSERT INTO tagged_data.active_prompts_b (b)
VALUES (%s)
""", [sum_b])
conn.commit()
conn.close()
```
