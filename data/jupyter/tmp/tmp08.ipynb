```python
import random

boxes = [[1,1], [1,0], [0,0]]

success = 0
total = 0

for i in range(10000):
    box = boxes[random.randrange(0, len(boxes))]
    ball_index = random.randrange(0, len(box))
    first_ball = box[ball_index]
    if first_ball == 1:
        total += 1
        box_copy = box.copy()
        del box_copy[ball_index]
        second_ball = box_copy[random.randrange(0, len(box_copy))]
        if second_ball == 1:
            success += 1

print(f"success: {success}, total: {total}, ratio: {round(success / total * 100)}")
```
