```python
from src import external_api
from src.db import DbConnection
survey_apps = external_api.read_sheet("1TIu0NXPH8LFgVmlzA4-y2GdI80MbnQW_-FAvAxDCvys", "survey apps", True)
survey_genre = external_api.read_sheet("1TIu0NXPH8LFgVmlzA4-y2GdI80MbnQW_-FAvAxDCvys", "survey genre", True)
survey_topic = external_api.read_sheet("1TIu0NXPH8LFgVmlzA4-y2GdI80MbnQW_-FAvAxDCvys", "survey topic", True)
```

```python
from src.db import DbConnection
conn = DbConnection()

values = [["title", "id"]]
for it in survey_apps:
    t = it["title"]
    p = f"%{t.lower()}%"
    rows = conn.select_all(f"""
    SELECT a.app_id
      FROM scraped_data.apps a
     WHERE LOWER(a.title) like %s
    """, [p])
    if len(rows) > 0:
        values.append([t, rows[0]["app_id"]])
    else:
        values.append([t, ""])

conn.close()
values
```

```python
for it in survey_apps:
    for g in survey_genre:
        for i in range(1, 6):
            if it["title"] == g[f"app{i}"]:
                it["genre"] = g["genre"]
                break
    for t in survey_topic:
        for i in range(1, 6):
            if it["title"] == t[f"app{i}"]:
                it["topic"] = t["topic"]
                break

values = [["title", "id", "genre", "topic"]]
apps = []
for it in survey_apps:
    if it["id"] in apps:
        continue
    apps.append(it["id"])
    values.append([it["title"], it["id"], it["genre"], it["topic"]])
values
```

```python
external_api.write_sheet("1TIu0NXPH8LFgVmlzA4-y2GdI80MbnQW_-FAvAxDCvys", "survey apps", values)

```
