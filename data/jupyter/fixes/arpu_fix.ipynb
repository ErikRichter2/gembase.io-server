```python
from src.runtime_constants import rr
from src.db.mysql import DbConnection
from src import external_api

sheet = external_api.read_sheet("1TIu0NXPH8LFgVmlzA4-y2GdI80MbnQW_-FAvAxDCvys", "arpu per node v2", True)

bulk_data = []
for row in sheet:
    tag_id_int = int(row["tag_id"])
    arpu = float(str(row["arpu"]).replace(",", "."))
    bulk_data.append((tag_id_int, arpu))

if len(bulk_data) > 0:
    conn = DbConnection(rr.ENV_PROD, is_remote=True)
    conn.query("""
    DELETE FROM platform.ext_arpu_per_tag
    """)
    conn.bulk("""
    INSERT INTO platform.ext_arpu_per_tag (tag_id_int, arpu) VALUES (%s, %s)
    """, bulk_data)
    conn.commit()
    conn.close()
    print("DONE")
```
