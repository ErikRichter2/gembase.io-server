```python
import numpy as np
rows = []
def processRow(r):
    app_id = r['Bundle Id GP']
    if app_id is None or app_id == 'None':
        app_id = r['Bundle Id Ios']
    if app_id is None or app_id == 'None':
        print(r['Title'])
        return
    d_gp = r['Downloads GP WW']
    d_ios = r['Downloads Ios WW']
    d = 0
    if not np.isnan(d_gp):
        d += d_gp
    if not np.isnan(d_ios):
        d += d_ios
    rows.append({
        'app_id': app_id,
        'title': r['Title'],
        'downloads': d
    })
df.apply(lambda r: processRow(r), axis=1)
rows
```

```python
from src.runtime_constants import rr
from src.db import DbConnection

rr.set_env(rr.ENV_DEV)
conn = DbConnection()
conn.query("""
DELETE FROM google_play_scraper.gpt_prompts
""")
for it in rows:
    for area in ['Genres', 'Mechanics', 'Content']:
        conn.query("""
        INSERT INTO google_play_scraper.gpt_prompts (app_id, title, area, downloads, status, requested, result) VALUES (%s, %s, %s, %s, 0, NULL, NULL)
        """, [it['app_id'], it['title'], area, it['downloads']])
    conn.commit()
conn.close()
```

```python
from src.runtime_constants import rr
from src.db import DbConnection

rr.set_env(rr.ENV_PROD)
conn = DbConnection()
rows = conn.select_all("""
SELECT google_id,real_installs,title FROM google_play_scraper.gps_games
""")

for row in rows:
    print(row['google_id'])
    cnt = conn.select_one("""
    SELECT count(1) as cnt FROM google_play_scraper.gpt_prompts WHERE app_id = %s
    """, [row['google_id']])
    if cnt['cnt'] == 0:
        for area in ['Genres', 'Mechanics', 'Content']:
            conn.query("""
            INSERT INTO google_play_scraper.gpt_prompts (app_id, title, area, downloads, status, requested, result, priority) VALUES (%s, %s, %s, %s, 0, NULL, NULL, 50)
            """, [row['google_id'], row['title'], area, row['real_installs']])
    else:
        conn.query("""
        UPDATE google_play_scraper.gpt_prompts SET priority = 50 WHERE app_id = %s
        """, [row['google_id']])

conn.commit()
conn.close()
```
